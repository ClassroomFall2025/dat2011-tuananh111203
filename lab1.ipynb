## Lab 1 - Bài 4

```python
# code lab 1 bài 4 ở đây
import math
a = float(input("Nhập hệ số a (a ≠ 0): "))
b = float(input("Nhập hệ số b: "))
c = float(input("Nhập hệ số c: "))
delta = b * b - 4 * a * c
print("delta=",abs(delta))
print("căn delta=",math.sqrt(abs(delta)))
```

## Lab 1 - Bài 5: bài tập làm thêm

```python
# code lab 1  bài làm thêm ở đây 
# phép tinh cơ bản
import math
a = 20
b = 10
print("cộng =", a + b)
print("trừ =", a - b)
print("nhân =", a * b)
print("chia =", a / b) 
#lũy thừa
x = 2
y = 3
print("lũy thừa =", x ** y)
#căn bậc hai
z = 16
print("căn bậc hai =", math.sqrt(z))
#hàm lượng giác
góc = 30
# chuyển góc sang radian   
radian = math.radians(góc)
print("sin(30) =", math.sin(radian))
print("cos(30) =", math.cos(radian))
print("tan(30) =", math.tan(radian))
#logarit
num = 100
print("log10(N) =", math.log10(num))
print("ln(N) =", math.log(num))
print("log10(N) =", math.log(num, 10))
```
