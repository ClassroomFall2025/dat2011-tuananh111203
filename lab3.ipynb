## Lab 3 - Bài 1

```python
# Code lab 3 bài 1 ở đây
# khai báo dãy số ban đầu chưa có phần tử
day_so = []
# khai báo biến đếm và tổng
dem = 0
tong = 0

# nhập dãy số
while True:
    so = input("Nhập số: ")
    if so.lower() == "q":
        print("thoat không nhập!")
        break
    else:
        try:
            phan_tu = int(so)
            day_so.append(phan_tu)
        except ValueError:
            print("Nhập không đúng, hãy nhập số hoặc q để thoát chương trình")

# sắp xếp dãy số tăng dần
day_so.sort(reverse=True)
print("dãy số đã sắp xếp tăng dần")
print(day_so)
print("Giá trị lơn nhất dãy số:", day_so[0])
print("Giá trị nhỏ nhất dãy số:", day_so[-1])

# duyệt dãy số và tính tổng dãy số chia hết cho 3
day_so = [2, 3, 6, 8, 9]

for phan_tu in day_so:
    if phan_tu % 3 == 0:
        tong += phan_tu
        dem += 1
        #xuất thông tin phần tử chia hết cho 3
        print("Phần tử chia hết cho 3:", phan_tu)
if dem != 0:
    print("giá trị cần tìm là:", tong/dem)
else:
    print("không có phần tử nào chia hết cho 3")
```

## Lab 3 - Bài 2

```python
# Code lab 3  bài 2 ở đây
danh_sach = []

while True:
    sv = {}
    sv["Họ tên"] = input("Nhập họ tên sinh viên: ")
    sv["Điểm"] = float(input("Nhập điểm của sinh viên: "))

    # Xếp loại
    diem = sv["Điểm"]
    if diem < 5:
        sv["Xếp loại"] = "Yếu"
    elif diem < 7:
        sv["Xếp loại"] = "Trung bình"
    elif diem < 8:
        sv["Xếp loại"] = "Khá"
    elif diem < 9:
        sv["Xếp loại"] = "Giỏi"
    else:
        sv["Xếp loại"] = "Xuất sắc"

    danh_sach.append(sv)

    tiep = input("Bạn có muốn nhập thêm sinh viên? (y/n): ")
    if tiep.lower() != "y":
        break

# Xuất danh sách
print("\n--- Danh sách sinh viên ---")
for sv in danh_sach:
    print(f"Họ tên: {sv['Họ tên']}, Điểm: {sv['Điểm']}, Xếp loại: {sv['Xếp loại']}")
```

## Lab 3 - Bài 3

```python
# Code lab 3  bài 3 ở đây
#Viết chương trình nhập một số nguyên từ bàn phím và cho biết số đó có phải là số nguyên tố hay không (số nguyên tố là số chỉ chia hết cho 1 và chính nó)
n = int(input("Nhập một số nguyên n: "))
if n < 2:
    print(n, " không phải là số nguyên tố ")
elif n == 2:
    print("2 là số nguyên tố.")
elif n % 2 == 0:
    print(n, " không phải là số nguyên tố ")   
else:
    ok = True
    i = 3
    while i * i <= n:
        if n % i == 0:
            ok = False
            break
        i += 2
    if ok:
        print(n, " là số nguyên tố ")
    else:
        print(n, " không phải là số nguyên tố ")
```

## Lab 3 - Bài 5: bài tập làm thêm

```python
# code lab 3  bài làm thêm ở đây 
print("---MÁY TÍNH ĐƠN GIẢN---")

while True:
  print("\n1. Cộng trừ (+)")
  print("2. Trừ (-)")
  print("3. Nhân (*)")
  print("4. Chia (/)")
  print("0. Thoát")

  lua_chon = input("Chọn phép tính: ")

  match lua_chon:
    case "0":
      print("BYE!")
      break
    case "1" | "+":
      a = float(input("Số thứ nhất: "))
      b = float(input("Số thứ hai: "))
      print(f"Kết quả: {a} + {b} = {a + b}")
    case "2" | "-":
      a = float(input("Số thứ nhất: "))
      b = float(input("Số thứ hai: "))
      print(f"Kết quả: {a} - {b} = {a - b}")
    case "3" | "*":
      a = float(input("Số thứ nhất: "))
      b = float(input("Số thứ hai: "))
      print(f"Kết quả: {a} * {b} = {a * b}")
    case "4" | "/":
      a = float(input("Số thứ nhất: "))
      b = float(input("Số thứ hai: "))
      match b:
        case 0:
          print("Không thể chia cho 0")
        case _:
          print(f"Kết quả: {a} / {b} = {a / b}")
    case _:
      print("Lựa chọn không hợp lệ")
```
