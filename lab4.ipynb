## Lab 1 - Bài 1

```python
# Code lab 4 bài 1 ở đây
def tinh_tien_nuoc(san_luong):
    """
    Hàm tính tiền nước sinh hoạt theo phương pháp lũy tiến.
    Tham số: san_luong (số nước sử dụng trong tháng, m³)
    """
    # Giá bán lẻ nước sạch sinh hoạt: Bậc 1: 7500, Bậc 2: 8800, Bậc 3: 12000, Bậc 4: 24000
    gia_ban_nuoc = (7500, 8800, 12000, 24000)
    tien_nuoc = 0
    
    # BẬC 1: <= 10m³
    if san_luong <= 10:
        tien_nuoc = san_luong * gia_ban_nuoc[0]
        
    # BẬC 2: Từ hơn 10m³ đến 20m³
    elif san_luong <= 20:
        # 10m³ đầu tiên theo giá Bậc 1, phần còn lại theo giá Bậc 2
        tien_nuoc = 10 * gia_ban_nuoc[0] + (san_luong - 10) * gia_ban_nuoc[1]
        
    # BẬC 3: Từ hơn 20m³ đến 30m³
    elif san_luong <= 30:
        # 10m³ Bậc 1 + 10m³ Bậc 2, phần còn lại theo giá Bậc 3
        tien_nuoc = 10 * gia_ban_nuoc[0] + 10 * gia_ban_nuoc[1] + (san_luong - 20) * gia_ban_nuoc[2]
        
    # BẬC 4: Hơn 30m³
    else: 
        # 10m³ Bậc 1 + 10m³ Bậc 2 + 10m³ Bậc 3, phần còn lại theo giá Bậc 4
        tien_nuoc = 10 * gia_ban_nuoc[0] + 10 * gia_ban_nuoc[1] + 10 * gia_ban_nuoc[2] + (san_luong - 30) * gia_ban_nuoc[3]
        
    return tien_nuoc

# --- PHẦN KIỂM TRA HÀM (Ví dụ) ---
# Yêu cầu đề bài: Tính tiền nước sinh hoạt cho 1 số trường hợp, ví dụ 12, 23, 34 m³
print("--- KIỂM TRA BÀI 1 ---")
print(f"Tiền nước cho 12 m³ là: {tinh_tien_nuoc(12):,.0f} đồng")
print(f"Tiền nước cho 23 m³ là: {tinh_tien_nuoc(23):,.0f} đồng")
print(f"Tiền nước cho 34 m³ là: {tinh_tien_nuoc(34):,.0f} đồng")


```

## Lab 4 - Bài 2

```python
# Code lab 4  bài 2 ở đây
def tinh_nguyen_lieu(sl_dau_xanh, sl_thap_cam, sl_deo):
    """
    Hàm tính tổng nguyên liệu (đường và đậu) cần thiết để làm 1 hộp bánh.
    Tham số: số lượng bánh đậu xanh, bánh thập cẩm, bánh dẻo.
    Kết quả trả về là một dictionary.
    """
    
    # 1. Định nghĩa tỷ lệ nguyên liệu (sử dụng tuple hoặc hằng số)
    # Tỷ lệ Đường: (Đậu xanh, Thập cẩm, Dẻo)
    DUONG_TY_LE = (0.04, 0.06, 0.05)
    # Tỷ lệ Đậu: (Đậu xanh, Thập cẩm, Dẻo)
    DAU_TY_LE = (0.07, 0, 0.02)
    
    # 2. Tính tổng lượng Đường theo công thức[cite: 40]:
    # Đường = SL_Đậu_xanh*0.04 + SL_Thập_cẩm*0.06 + SL_Dẻo*0.05
    tong_duong = (sl_dau_xanh * DUONG_TY_LE[0] +
                  sl_thap_cam * DUONG_TY_LE[1] +
                  sl_deo * DUONG_TY_LE[2])
    
    # 3. Tính tổng lượng Đậu theo công thức[cite: 41]:
    # Đậu = SL_Đậu_xanh*0.07 + SL_Thập_cẩm*0 + SL_Dẻo*0.02
    tong_dau = (sl_dau_xanh * DAU_TY_LE[0] +
                sl_thap_cam * DAU_TY_LE[1] +
                sl_deo * DAU_TY_LE[2])
    
    # 4. Trả về kết quả dưới dạng dictionary (ví dụ: {"sugar": 1, "bean": 2}) [cite: 42]
    nguyen_lieu_can_thiet = {
        "Duong (kg)": round(tong_duong, 3), # Làm tròn 3 chữ số thập phân
        "Dau (kg)": round(tong_dau, 3)
    }
    
    return nguyen_lieu_can_thiet

# ----------------------------------------------------
# KIỂM TRA HÀM (Ví dụ)
print("--- KẾT QUẢ TÍNH NGUYÊN LIỆU LÀM BÁNH ---")

# Giả sử làm 1 hộp có: 5 bánh đậu xanh, 2 bánh thập cẩm, 3 bánh dẻo
sl_dau_xanh = 5
sl_thap_cam = 2
sl_deo = 3

nguyen_lieu_hop_banh = tinh_nguyen_lieu(sl_dau_xanh, sl_thap_cam, sl_deo)

print(f"Số lượng bánh: ĐX={sl_dau_xanh}, TC={sl_thap_cam}, D={sl_deo}")
print("Tổng nguyên liệu cần thiết:")
print(nguyen_lieu_hop_banh)

# ----------------------------------------------------
# Ví dụ tính toán (Kiểm tra):
# Đường = (5 * 0.04) + (2 * 0.06) + (3 * 0.05) = 0.20 + 0.12 + 0.15 = 0.47 kg
# Đậu = (5 * 0.07) + (2 * 0) + (3 * 0.02) = 0.35 + 0.00 + 0.06 = 0.41 kg
```

## Lab 4 - Bài 3

```python
# Code lab 4  bài 3 ở đây
def loc_so_chan():
    """
    Hàm nhập một dãy số nguyên (nhập 's' để dừng)
    và lọc ra các số chẵn bằng cách sử dụng hàm lambda và filter.
    Kèm theo xử lý ngoại lệ cho việc nhập liệu.
    """
    my_list = []
    print("\n--- 3. LỌC SỐ CHẴN TRONG DÃY SỐ ---")
    print("Nhập dãy số nguyên (nhấn Enter sau mỗi số, nhập 's' hoặc 'S' để dừng):")
    
    while True:
        # Nhận dữ liệu nhập vào
        nhap_lieu = input("> ")
        
        # Yêu cầu nâng cao: Dừng khi nhập 's' hoặc 'S'
        if nhap_lieu.lower() == 's':
            break
            
        try:
            # Chuyển đổi input sang số nguyên
            so = int(nhap_lieu)
            my_list.append(so)
        except ValueError:
            # Quản lý ngoại lệ: Thông báo lỗi nếu input không phải số nguyên
            print("❌ Lỗi: Vui lòng chỉ nhập SỐ NGUYÊN. (Hoặc 's' để dừng)")
            continue
            
    if not my_list:
        print("Danh sách rỗng, không có số nào để lọc.")
        return
        
    # SỬ DỤNG HÀM LAMBDA VÀ FILTER
    # lambda x: (x % 2 == 0) kiểm tra xem số x có chia hết cho 2 (là số chẵn) không.
    # filter() áp dụng lambda này cho từng phần tử trong my_list.
    # list() chuyển đổi kết quả filter (một object filter) thành danh sách.
    so_chan = list(filter(lambda x: (x % 2 == 0), my_list))
    
    # In ra kết quả
    print("\n--------------------------------")
    print(f"Dãy số đã nhập: {my_list}")
    print(f"Các số chẵn được lọc: {so_chan}")
    print("--------------------------------")

# --- CHẠY THỬ HÀM ---
# Nếu bạn chạy trực tiếp file này (ví dụ: python lab4_functions_bai3.py)
# thì hàm sẽ được gọi để kiểm tra.
if __name__ == "__main__":
    loc_so_chan()
```

## Lab 4 - Bài 4

```python
# code lab 4 bài 4 ở đây
# ==========================================================
# FILE: lab4_main.py
# CHỨC NĂNG: Tổ chức Menu chính, gọi các hàm từ mô-đun lab4_functions
# ĐÁP ỨNG YÊU CẦU BÀI 4 (LAB 4)
# ==========================================================

# SỬ DỤNG LỆNH IMPORT: Khai báo sử dụng mô-đun chứa các hàm đã viết
import lab4 as lf

def hien_thi_menu():
    """Hiển thị thực đơn (menu) theo mẫu đề bài (thêm Bài 3)."""
    print("\n" + "="*10 + " MENU CHƯƠNG TRÌNH " + "="*10)
    print("1. Tính tiền nước sinh hoạt (Bài 1)")
    print("2. Tính nguyên liệu làm bánh (Bài 2)")
    print("3. Lọc số chẵn trong dãy số (Bài 3)")
    print("4. Kết thúc")
    print("="*38)

def thuc_hien_chuc_nang(chon):
    """Xử lý logic cho từng chức năng đã chọn."""
    if chon == 1:
        print("\n--- 1. TÍNH TIỀN NƯỚC SINH HOẠT ---")
        try:
            # Quản lý ngoại lệ cho input của hàm
            san_luong = float(input("Nhập số nước đã sử dụng (m³): "))
            # Gọi hàm Bài 1
            tien = lf.tinh_tien_nuoc(san_luong)
            print(f"✅ Tổng tiền nước cần thanh toán là: {tien:,.0f} đồng")
        except ValueError:
            print("❌ Lỗi: Vui lòng nhập một số hợp lệ cho sản lượng.")

    elif chon == 2:
        print("\n--- 2. TÍNH NGUYÊN LIỆU LÀM BÁNH ---")
        try:
            # Quản lý ngoại lệ cho input của hàm
            dz = int(input("Nhập số bánh đậu xanh: "))
            tc = int(input("Nhập số bánh thập cẩm: "))
            d = int(input("Nhập số bánh dẻo: "))
            
            # Gọi hàm Bài 2
            nl = lf.tinh_nguyen_lieu(dz, tc, d)
            tong_banh = dz + tc + d
            print(f"✅ Nguyên liệu cần thiết để làm {tong_banh} cái bánh:")
            # Lấy giá trị từ dictionary trả về
            print(f"- Đường: {nl['Đường (kg)']} kg")
            print(f"- Đậu: {nl['Đậu (kg)']} kg")
        except ValueError:
            print("❌ Lỗi: Vui lòng nhập số lượng bánh là số nguyên.")
    
    elif chon == 3:
         # Gọi hàm Bài 3 (Hàm này đã có logic nhập liệu và xử lý lỗi bên trong)
         lf.loc_so_chan()
    
def main():
    """Chức năng chính của chương trình, chứa vòng lặp menu."""
    while True:
        hien_thi_menu()
        
        try:
            # Nhận vào 1 số từ 1 đến 4 (Quản lý ngoại lệ)
            chon = int(input("Chọn chức năng: "))
            
            if chon == 4:
                print("👋 Ứng dụng kết thúc. Tạm biệt!")
                break
            
            # Thông báo lỗi nếu số có giá trị ngoài khoảng từ 1 đến 4
            if chon < 1 or chon > 4:
                print("❌ Lỗi: Chức năng không hợp lệ. Vui lòng chọn số từ 1 đến 4.")
                continue

            # Sử dụng lệnh if/elif để gọi đến các hàm chức năng tương ứng
            thuc_hien_chuc_nang(chon)

        except ValueError:
            # Thông báo lỗi nếu người dùng không nhập vào số
            print("❌ Lỗi: Vui lòng chỉ nhập số để chọn chức năng.")

if __name__ == "__main__":
    main()
```
